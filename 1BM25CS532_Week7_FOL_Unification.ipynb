```python
def is_variable(expr):
    return isinstance(expr, str) and expr.islower() and len(expr) == 1

def occurs_check(var, expr, subs):
    if var == expr:
        return True
    elif is_variable(expr) and expr in subs:
        return occurs_check(var, subs[expr], subs)
    elif isinstance(expr, list):
        return any(occurs_check(var, sub_expr, subs) for sub_expr in expr)
    return False

def unify_var(var, x, subs):

    if var in subs:
        return unify(subs[var], x, subs)
    elif is_variable(x) and x in subs:
        return unify(var, subs[x], subs)
    elif occurs_check(var, x, subs):
        return None  # Occurs check failed
    else:
        new_subs = subs.copy()
        new_subs[var] = x
        return new_subs

def unify(expr1, expr2, subs=None):
    if subs is None:
        subs = {}

    if subs is None:
        return None

    if expr1 == expr2:
        return subs

    elif is_variable(expr1):
        return unify_var(expr1, expr2, subs)
    elif is_variable(expr2):
        return unify_var(expr2, expr1, subs)

    elif isinstance(expr1, list) and isinstance(expr2, list):
        if len(expr1) != len(expr2):
            return None
        subs = unify(expr1[0], expr2[0], subs)
        for arg1, arg2 in zip(expr1[1:], expr2[1:]):
            subs = unify(arg1, arg2, subs)
            if subs is None:
                return None
        return subs

    # Case 4: Mismatch
    else:
        return None

# Driver code with test cases
if __name__ == "__main__":
    print("--- FOL UNIFICATION ALGORITHM ---\n")

    # Test Case 1: P(x, f(g(x))) and P(A, y)
    # Represented as lists: ['P', 'x', ['f', ['g', 'x']]]
    expr1 = ['P', 'x', ['f', ['g', 'x']]]
    expr2 = ['P', 'A', 'y']

    print(f"Expression 1 : {expr1}")
    print(f"Expression 2 : {expr2}")
    result = unify(expr1, expr2)

    if result is not None:
        print(f"Result : Unified Successfully!")
        print(f"Substitution : {result}\n")
    else:
        print("Result       : Unification Failed!\n")

    expr3 = ['Knows', 'John', 'x']
    expr4 = ['Knows', 'y', 'Jane']

    print(f"Expression 3 : {expr3}")
    print(f"Expression 4 : {expr4}")
    result2 = unify(expr3, expr4)

    if result2 is not None:
        print(f"Result : Unified Successfully!")
        print(f"Substitution : {result2}\n")
    else:
        print("Result : Unification Failed!\n")

    expr5 = ['P', 'x']
    expr6 = ['P', ['f', 'x']]

    print(f"Expression 5 : {expr5}")
    print(f"Expression 6 : {expr6}")
    result3 = unify(expr5, expr6)

    if result3 is not None:
        print(f"Result : Unified Successfully!")
        print(f"Substitution : {result3}\n")
    else:
        print("Result : Unification Failed (Occurs Check)!\n")
```
