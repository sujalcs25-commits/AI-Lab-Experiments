```python
import heapq

GOAL = (1, 2, 3, 4, 5, 6, 7, 8, 0)

class State:
    def __init__(self, board, g=0, parent=None, h_type="manhattan"):
        self.board, self.g, self.parent, self.h_type = tuple(board), g, parent, h_type
        self.h = self.calc_h()
        self.f = self.g + self.h

    def calc_h(self):
        if self.h_type == "misplaced":
            return sum(1 for i in range(9) if self.board[i] and self.board[i] != GOAL[i])
        return sum(
            abs(i // 3 - GOAL.index(val) // 3) + abs(i % 3 - GOAL.index(val) % 3)
            for i, val in enumerate(self.board) if val
        )

    def get_neighbors(self):
        idx = self.board.index(0)
        r, c = divmod(idx, 3)
        neighbors = []
        for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
            nr, nc = r + dr, c + dc
            if 0 <= nr < 3 and 0 <= nc < 3:
                n_idx = nr * 3 + nc
                b = list(self.board)
                b[idx], b[n_idx] = b[n_idx], b[idx]
                neighbors.append(State(b, self.g + 1, self, self.h_type))
        return neighbors

    def __lt__(self, other):
        return self.f < other.f

def solve(start_board, h_type):
    pq, visited = [State(start_board, h_type=h_type)], set()
    while pq:
        curr = heapq.heappop(pq)
        if curr.board == GOAL:
            path = []
            while curr:
                path.append(curr)
                curr = curr.parent
            return path[::-1]
        visited.add(curr.board)
        for nxt in curr.get_neighbors():
            if nxt.board not in visited:
                heapq.heappush(pq, nxt)
    return None

def print_path(path, title):
    print(f"\n=== {title} ===")
    if not path:
        print("Unsolvable puzzle!")
        return
    for step, node in enumerate(path):
        print(f"Step {step} | g(n)={node.g}, h(n)={node.h}, f(n)={node.f}")
        b = node.board
        print(f" {b[0]} {b[1]} {b[2]}\n {b[3]} {b[4]} {b[5]}\n {b[6]} {b[7]} {b[8]}\n")

start = tuple(map(int, input("Enter 9 board numbers (0-8 space-separated): ").split()))


path_misplaced = solve(start, "misplaced")
print_path(path_misplaced, "Case 1: Misplaced Tiles")


path_manhattan = solve(start, "manhattan")
print_path(path_manhattan, "Case 2: Manhattan Distance")
```
