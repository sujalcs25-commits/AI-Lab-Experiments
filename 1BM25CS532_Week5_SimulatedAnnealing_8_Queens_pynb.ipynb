```python
import random
import math

def calculate_cost(board):
    h = 0
    n = len(board)
    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                h += 1
    return h

def print_board(board):
    n = len(board)
    print("+" + "---+" * n)
    for r in range(n):
        row_str = "|"
        for c in range(n):
            if board[c] == r:
                row_str += " Q |"
            else:
                row_str += " . |"
        print(row_str)
        print("+" + "---+" * n)

def simulated_annealing(n=8, temp=1000.0, cooling_rate=0.99):
    current = [random.randint(0, n - 1) for _ in range(n)]
    step = 1

    while temp > 0.001:
        current_cost = calculate_cost(current)
        print(f"\nStep {step} State: {current} (Cost: {current_cost}, Temp: {temp:.2f})")
        print_board(current)

        if current_cost == 0:
            return current, "Goal Reached"

        col = random.randint(0, n - 1)
        row = random.randint(0, n - 1)
        neighbor = list(current)
        neighbor[col] = row

        neighbor_cost = calculate_cost(neighbor)
        delta_E = neighbor_cost - current_cost

        if delta_E < 0 or random.random() < math.exp(-delta_E / temp):
            current = neighbor

        temp *= cooling_rate
        step += 1

    return current, "Terminated (Max iterations reached)"

solution, status = simulated_annealing(8)
print(f"\nStatus: {status}")
print(f"Final Board State: {solution}")
print(f"Attacking Pairs (Cost): {calculate_cost(solution)}")
print_board(solution)
```
