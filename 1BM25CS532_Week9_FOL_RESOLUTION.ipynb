```python
from nltk.sem import Expression
from nltk.inference import ResolutionProver

read_expr = Expression.fromstring

# Resolution function
def resolution(KB, goal):
    clauses = [read_expr(statement) for statement in KB]
    query = read_expr(goal)

    return ResolutionProver().prove(
        query, clauses, verbose=True
    )


# Case Study 1: Leela and Oshin
KB1 = [
    "Mother(leela, oshin)",
    "Alive(leela)",
    "all x y. (Mother(x,y) -> Parent(x,y))",
    "all x y. ((Parent(x,y) & Alive(x)) -> Older(x,y))"
]

goal1 = "Older(leela, oshin)"

print("Case Study 1: Leela and Oshin")
result1 = resolution(KB1, goal1)
print("Query:", goal1)
print("Result:", result1)


# Case Study 2: It will rain
KB2 = [
    "Humidity(high) | Cloudy(sky)",
    "Cloudy(sky) -> Rain(today)",
    "Humidity(high) -> Hot(today)",
    "-Hot(today)"
]

goal2 = "Rain(today)"

print("\nCase Study 2: It will rain")
result2 = resolution(KB2, goal2)
print("Query:", goal2)
print("Result:", result2)
```
