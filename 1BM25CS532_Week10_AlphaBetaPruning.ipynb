```python

visited = []

def alpha_beta(node, alpha, beta, maximizing):

    # Leaf node
    if isinstance(node, int):
        visited.append(node)
        return node

    # MAX node
    if maximizing:
        best = float('-inf')

        for child in node:
            value = alpha_beta(child, alpha, beta, False)
            best = max(best, value)
            alpha = max(alpha, best)

            if alpha >= beta:
                break

        return best

    # MIN node
    else:
        best = float('inf')

        for child in node:
            value = alpha_beta(child, alpha, beta, True)
            best = min(best, value)
            beta = min(beta, best)

            if alpha >= beta:
                break

        return best


# Given game tree
tree = [
    [[4, 3], [6, 2]],
    [[2, 1], [3, 5], [3, 1]],
    [[5, 4], [7, 5]]
]

result = alpha_beta(
    tree, float('-inf'), float('inf'), True
)

print("Optimal value:", result)
print("Visited leaves:", visited)
print("Total leaves:", 14)
print("Evaluated leaves:", len(visited))
print("Pruned leaves:", 14 - len(visited))
```
