```python
import random

def calculate_cost(board):
    h = 0
    n = len(board)
    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                h += 1
    return h

def print_board(board):
    n = len(board)
    print("+" + "---+" * n)
    for r in range(n):
        row_str = "|"
        for c in range(n):
            if board[c] == r:
                row_str += " Q |"
            else:
                row_str += " . |"
        print(row_str)
        print("+" + "---+" * n)

def hill_climbing(n=4):
    current = [random.randint(0, n - 1) for _ in range(n)]
    step = 1

    while True:
        current_cost = calculate_cost(current)
        print(f"\nStep {step} State: {current} (Cost: {current_cost})")
        print_board(current)

        if current_cost == 0:
            return current, "Goal Reached"

        best_neighbor = None
        best_cost = current_cost

        for col in range(n):
            for row in range(n):
                if current[col] != row:
                    neighbor = list(current)
                    neighbor[col] = row
                    neighbor_cost = calculate_cost(neighbor)

                    if neighbor_cost < best_cost:
                        best_cost = neighbor_cost
                        best_neighbor = neighbor

        if best_cost >= current_cost:
            return current, "Stuck in Local Optimum"

        current = best_neighbor
        step += 1

solution, status = hill_climbing(4)
print(f"\nStatus: {status}")
print(f"Final Board State: {solution}")
print(f"Attacking Pairs (Cost): {calculate_cost(solution)}")
print_board(solution)
```
